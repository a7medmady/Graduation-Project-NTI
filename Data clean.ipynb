```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("ecommerce_customer_churn_dataset.csv")

df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
missing = df.isnull().sum()

missing = missing[missing > 0].sort_values(ascending=False)

print(missing)
```

```python
duplicates = df.duplicated().sum()

print("Number of duplicated rows:", duplicates)
```

```python
df = df.drop_duplicates().reset_index(drop=True)
```

```python
text_columns = df.select_dtypes(include="object").columns

for col in text_columns:
    df[col] = df[col].str.strip()
```

```python
df["Gender"] = df["Gender"].str.title()
df["Country"] = df["Country"].str.strip()
df["City"] = df["City"].str.title()
df["Signup_Quarter"] = df["Signup_Quarter"].str.upper().str.strip()
```

```python
for col in text_columns:
    print(col, df[col].unique())
```

```python
df.loc[(df["Age"] < 18) | (df["Age"] > 100), "Age"] = np.nan
```

```python
df.loc[
    (df["Cart_Abandonment_Rate"] < 0) |
    (df["Cart_Abandonment_Rate"] > 100),
    "Cart_Abandonment_Rate"
] = np.nan
```

```python
df.loc[df["Total_Purchases"] < 0, "Total_Purchases"] = np.nan
```

```python
df.loc[
    (df["Discount_Usage_Rate"] < 0) |
    (df["Discount_Usage_Rate"] > 100),
    "Discount_Usage_Rate"
] = np.nan
```

```python
df.loc[
    (df["Returns_Rate"] < 0) |
    (df["Returns_Rate"] > 100),
    "Returns_Rate"
] = np.nan
```

```python
df.loc[
    (df["Email_Open_Rate"] < 0) |
    (df["Email_Open_Rate"] > 100),
    "Email_Open_Rate"
] = np.nan
```

```python
df.loc[
    (df["Social_Media_Engagement_Score"] < 0) |
    (df["Social_Media_Engagement_Score"] > 100),
    "Social_Media_Engagement_Score"
] = np.nan
```

```python
df.loc[
    (df["Mobile_App_Usage"] < 0) |
    (df["Mobile_App_Usage"] > 100),
    "Mobile_App_Usage"
] = np.nan
```

```python
non_negative_columns = [
    "Membership_Years",
    "Login_Frequency",
    "Session_Duration_Avg",
    "Pages_Per_Session",
    "Wishlist_Items",
    "Days_Since_Last_Purchase",
    "Customer_Service_Calls",
    "Product_Reviews_Written",
    "Lifetime_Value",
    "Credit_Balance"
]

for col in non_negative_columns:
    df.loc[df[col] < 0, col] = np.nan
```

```python
df.loc[
    (df["Payment_Method_Diversity"] < 1) |
    (df["Payment_Method_Diversity"] > 5),
    "Payment_Method_Diversity"
] = np.nan
```

```python
numeric_columns = df.select_dtypes(include=np.number).columns

numeric_columns
```

```python
for col in numeric_columns:
    if col != "Churned":
        df[col] = df[col].fillna(df[col].median())
```

```python
categorical_columns = df.select_dtypes(include="object").columns

for col in categorical_columns:
    df[col] = df[col].fillna(df[col].mode()[0])
```

```python
integer_columns = [
    "Age",
    "Login_Frequency",
    "Wishlist_Items",
    "Customer_Service_Calls",
    "Product_Reviews_Written",
    "Payment_Method_Diversity",
    "Churned"
]

for col in integer_columns:
    df[col] = df[col].round().astype(int)
```

```python
print("Total missing values:", df.isnull().sum().sum())
```

```python
print("Total duplicated rows:", df.duplicated().sum())
```
